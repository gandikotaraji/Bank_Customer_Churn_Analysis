```python
numeric_df = df.select_dtypes(include='number')
df['score'] = df.groupby(['estimated_salary', 'credit_score'])[numeric_df.columns].transform(lambda x: x.fillna(x.median())).iloc[:, 0]
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
#create a column age_group
labels=['Young adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column of customers for credit card status

credit_card_mapping = {
    0: 'No Credit Card',
    1: 'Has Credit Card'
}

df['credit_card_status'] = df['credit_card'].map(credit_card_mapping)
```
